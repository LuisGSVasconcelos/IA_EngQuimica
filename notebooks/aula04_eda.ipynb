## 3.1 — Exercício Guiado: Roteiro EDA de 6 Passos

Siga passo-a-passo as células abaixo.

### Passo 1: Importar bibliotecas

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
```

### Passo 2: Carregar o dataset + Visão Geral

```python
URL = "https://raw.githubusercontent.com/LuisGSVasconcelos/IA_EngQuimica/main/dados/aula04/coluna_destilacao_30dias.csv"

df = pd.read_csv(URL, parse_dates=['timestamp'])
df.set_index('timestamp', inplace=True)

print(f"Shape: {df.shape}")
print(f"\nInfo:")
print(df.info())
print(f"\nDescriptive stats:")
print(df.describe().round(2))
```

### Passo 3: Histogramas (Univariada)

```python
cols = ['T_top_C', 'T_base_C', 'P_coluna_kPa', 'composicao_destilado']
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.ravel(), cols):
    sns.histplot(df[col], bins=50, kde=True, ax=ax, color='steelblue')
    ax.set_title(f'Distribuição de {col}')
plt.tight_layout()
plt.show()
```

### Passo 4: Boxplot

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=df[cols], orient='h', palette='Set2')
plt.title('Boxplot das variáveis principais')
plt.tight_layout()
plt.show()
```

### Passo 5: Séries Temporais

```python
# Amostra de 3 dias
df_3d = df['2026-08-01':'2026-08-03']
df_3d[['T_top_C', 'T_base_C']].plot(figsize=(14, 5), title='Temperaturas — 3 dias')
plt.ylabel('T (°C)')
plt.tight_layout()
plt.show()
```

### Passo 6: Pairplot + Heatmap

```python
# Pairplot (amostra para não travar — 2000 pontos aleatórios)
df_sample = df.sample(2000, random_state=42)
sns.pairplot(df_sample, vars=cols, corner=True, kind='reg',
             plot_kws={'line_kws': {'color': 'red', 'alpha': 0.3}})
plt.suptitle('Pairplot — Variáveis Principais', y=1.02)
plt.show()
```

```python
# Heatmap de correlação
plt.figure(figsize=(10, 8))
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, fmt='.2f', cmap='RdBu', center=0,
            square=True, linewidths=0.5)
plt.title('Matriz de Correlação — Coluna de Destilação')
plt.tight_layout()
plt.show()
```
