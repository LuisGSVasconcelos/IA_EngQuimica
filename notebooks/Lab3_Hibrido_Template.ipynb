## Balanço de Massa (est. estacionário)

$$C_A = \frac{F \cdot C_{A0}}{F + V \cdot k}$$

Isolando $k$:  $$k = \frac{F \cdot (C_{A0} - C_A)}{V \cdot C_A}$$
