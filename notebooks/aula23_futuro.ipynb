## 3.1 — Autoavaliação de Competências (Radar)


### Passo 1: defina seu perfil (1-5)


```python
# Edite os valores SEGUNDO sua autoavaliacao honesta
perfil = {
    'Fund. EQ':        3,
    'Python/Dados':    3,
    'ML/IA':           3,
    'Controle/Otim.':  3,
    'Soft Skills':     4,
    'Dominio Industrial': 3,
}
print('Confianca:', {k: '5 - especialista' if v>=4 else ('3 - avançando' if v>=3 else '1 - iniciante') for k,v in perfil.items()})
```

### Passo 2: plote seu radar (grafico de teia)


```python
import numpy as np
import matplotlib.pyplot as plt

labels = list(perfil.keys())
n = len(labels)
ang = np.linspace(0, 2*np.pi, n, endpoint=False).tolist()
ang += ang[:1]
v = list(perfil.values()) + [perfil[labels[0]]]

ax = plt.subplot(111, polar=True)
ax.fill(ang, v, alpha=0.35)
ax.plot(ang, v, lw=2.5)
ax.set_xticks(ang[:-1])
ax.set_xticklabels(labels)
ax.set_ylim(0, 5)
plt.title(f'Média = {np.mean(list(perfil.values())):.1f}/5')
plt.show()
```

### Passo 3: identifique forças e lacunas


```python
forcas = {k: v for k, v in perfil.items() if v >= 4}
lacunas = {k: v for k, v in perfil.items() if v <= 2}
print('Forças:', forcas or 'nenhuma ainda')
print('Lacunas:', lacunas or 'nenhuma')
print()
print('Pergunta: qual lacuna é a MAIS ESTRATÉGICA para fechar nos próximos 12 meses?')
```
